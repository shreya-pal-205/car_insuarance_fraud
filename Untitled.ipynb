```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
```

```python
df = pd.read_csv("file(3).csv")
print("Shape: " ,df.shape)
df.head()
```

```python
# %%
# Quick info and target check
print(df.columns.tolist())
print(df.info())
print(df['CLAIM_FLAG'].value_counts(dropna=False))
# If CLAIM_FLAG is 'Y'/'N', convert:
if df['CLAIM_FLAG'].dtype == object:
    df['CLAIM_FLAG'] = df['CLAIM_FLAG'].map({'Y':1, 'N':0})
```

```python
# %%
# 2) Basic data cleaning (modify as needed)
# - Drop ID column if present
if 'ID' in df.columns:
    df = df.drop(columns=['ID'])
# - Trim whitespace from object columns
for c in df.select_dtypes(include='object').columns:
    df[c] = df[c].str.strip()

# Show missing value summary
missing = df.isnull().sum().sort_values(ascending=False)
missing[missing>0]
```

```python
# %%
# Strategy for missing values (examples: median for numeric, mode for categorical)
num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()
print("Numeric cols:", num_cols)
print("Categorical cols:", cat_cols)

# Fill numerics with median, categoricals with mode
for c in num_cols:
    df[c] = df[c].fillna(df[c].median())
for c in cat_cols:
    df[c] = df[c].fillna(df[c].mode().iloc[0])
```

```python
# %%
# 3) Quick EDA: target balance and pie chart
plt.figure(figsize=(6,6))
counts = df['CLAIM_FLAG'].value_counts()
plt.pie(counts, labels=['Not Fraud','Fraud'], autopct='%1.1f%%', startangle=90)
plt.title('Fraud vs Not-Fraud Distribution')
plt.show()

print("Counts:\n", counts)

```

```python
# 4) Feature engineering basics (you can expand this)

# Convert income and claim amount columns to numeric (if they aren't already)
for col in ['CLM_AMT', 'INCOME']:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')  # convert non-numeric to NaN

# Example: create a ratio of claims to income
if 'CLM_AMT' in df.columns and 'INCOME' in df.columns:
    df['claim_income_ratio'] = df['CLM_AMT'] / (df['INCOME'] + 1)

# Convert binary-like text columns to 0/1 if present
binary_map = {'Yes': 1, 'No': 0, 'Y': 1, 'N': 0, 'yes': 1, 'no': 0, 'y': 1, 'n': 0}

for c in df.columns:
    if df[c].dtype == 'object':  # only for object columns
        unique_vals = set(df[c].dropna().unique())
        if unique_vals.issubset(set(binary_map.keys())):
            df[c] = df[c].map(binary_map)
```

```python
# %%
# 5) Prepare X, y
target = 'CLAIM_FLAG'
X = df.drop(columns=[target])
y = df[target].astype(int)

# Keep a copy of original features for later visualization
X_orig = X.copy()
```

```python
# %%
# 6) Preprocessing: identify categorical vs numerical cols for pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

numeric_features = X.select_dtypes(include=[np.number]).columns.tolist()
categorical_features = X.select_dtypes(include=['object', 'category']).columns.tolist()

print("Numeric:", numeric_features)
print("Categorical:", categorical_features)

# Numeric transformer: median impute + standard scaler
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

# Categorical transformer: most frequent impute + one-hot
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

preprocessor = ColumnTransformer(transformers=[
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
])
```

```python
# %%
# 7) Split into train/test sets
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, stratify=y, random_state=42)
print(X_train.shape, X_test.shape, y_train.value_counts(normalize=True))
```

```python
# %%
# 8) Handle class imbalance on training set using SMOTE (after preprocessing)
# We'll build a pipeline that does preprocessing -> SMOTE -> classifier using imbalanced-learn
try:
    from imblearn.pipeline import Pipeline as ImbPipeline
    from imblearn.over_sampling import SMOTE
except Exception as e:
    print("Install imbalanced-learn to use SMOTE: pip install imbalanced-learn")
    raise e

smote = SMOTE(random_state=42)
```

```python
# %%
# 9) Define candidate models
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, roc_curve, accuracy_score

# Try to import XGBoost and LightGBM if available
use_xgb = False
use_lgb = False
try:
    import xgboost as xgb
    from xgboost import XGBClassifier
    use_xgb = True
except:
    print("XGBoost not installed — skipping XGBoost (pip install xgboost)")

try:
    import lightgbm as lgb
    from lightgbm import LGBMClassifier
    use_lgb = True
except:
    print("LightGBM not installed — skipping LightGBM (pip install lightgbm)")
```

```python
# %%
# We'll prepare a dict of models
models = {
    "LogisticRegression": LogisticRegression(max_iter=1000, class_weight='balanced', random_state=42),
    "RandomForest": RandomForestClassifier(n_estimators=200, class_weight='balanced', random_state=42),
    "SVM": SVC(probability=True, class_weight='balanced', random_state=42),
    "KNN": KNeighborsClassifier(n_neighbors=5)
}
if use_xgb:
    models["XGBoost"] = XGBClassifier(use_label_encoder=False, eval_metric='logloss', random_state=42)
if use_lgb:
    models["LightGBM"] = LGBMClassifier(random_state=42)
```

```python
# %%
# 10) Train each model using pipeline: preprocessor -> SMOTE -> classifier
from collections import defaultdict
results = defaultdict(dict)

for name, clf in models.items():
    print(f"\nTraining: {name}")
    pipe = ImbPipeline(steps=[
        ('preprocessor', preprocessor),
        ('smote', SMOTE(random_state=42)),
        ('clf', clf)
    ])
    pipe.fit(X_train, y_train)
    # Save the pipeline temporarily in results to evaluate
    y_pred = pipe.predict(X_test)
    y_proba = pipe.predict_proba(X_test)[:,1] if hasattr(pipe, "predict_proba") else None
    acc = accuracy_score(y_test, y_pred)
    auc = roc_auc_score(y_test, y_proba) if y_proba is not None else None
    print("Accuracy:", round(acc,4))
    if auc is not None:
        print("ROC-AUC:", round(auc,4))
    print(classification_report(y_test, y_pred, digits=4))
    results[name]['pipeline'] = pipe
    results[name]['accuracy'] = acc
    results[name]['roc_auc'] = auc
```

```python
# %%
# 11) Choose best model (by ROC-AUC if available else accuracy)
best_model_name = None
best_score = -1
for name, info in results.items():
    score = info['roc_auc'] if info['roc_auc'] is not None else info['accuracy']
    print(name, "score:", score)
    if score is not None and score > best_score:
        best_score = score
        best_model_name = name

print("\nBest model:", best_model_name, "with score:", best_score)
best_pipeline = results[best_model_name]['pipeline']
```

```python
# %%
# 12) Confusion matrix and heatmap for best model
y_pred_best = best_pipeline.predict(X_test)
cm = confusion_matrix(y_test, y_pred_best)
plt.figure(figsize=(6,4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Not Fraud','Fraud'], yticklabels=['Not Fraud','Fraud'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title(f'Confusion Matrix - {best_model_name}')
plt.show()
```

```python
# %%
# 13) ROC Curve (if probability available)
if hasattr(best_pipeline, "predict_proba"):
    y_proba_best = best_pipeline.predict_proba(X_test)[:,1]
    fpr, tpr, _ = roc_curve(y_test, y_proba_best)
    plt.figure(figsize=(6,4))
    plt.plot(fpr, tpr, label=f'{best_model_name} (AUC={roc_auc_score(y_test, y_proba_best):.3f})')
    plt.plot([0,1],[0,1],'k--')
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title('ROC Curve')
    plt.legend()
    plt.show()
```

```python
# %%
# 15) Additional visualizations: target by some categorical columns if exist
for c in ['OCCUPATION','CAR_USE','MSTATUS']:
    if c in df.columns:
        plt.figure(figsize=(8,4))
        tmp = pd.crosstab(df[c], df['CLAIM_FLAG'], normalize='index')*100
        tmp.plot(kind='bar', stacked=True, figsize=(8,4))
        plt.ylabel('Percentage')
        plt.title(f'Fraud% by {c}')
        plt.legend(['Not Fraud','Fraud'])
        plt.show()
```

```python
# %%
# 17) Simple interactive test function (CLI-style) — you can run this cell and enter values
def predict_from_input(pipeline):
    print("Enter values for the following features. Press Enter to use default (if shown).")
    sample = {}
    for col in X.columns:
        default = X[col].median() if X[col].dtype != 'object' else X[col].mode().iloc[0]
        val = input(f"{col} (default {default}): ")
        if val.strip()=="":
            sample[col] = default
        else:
            # try convert to numeric if original column numeric
            if X[col].dtype != 'object':
                try:
                    sample[col] = float(val)
                except:
                    sample[col] = default
            else:
                sample[col] = val
    sample_df = pd.DataFrame([sample])
    pred = pipeline.predict(sample_df)[0]
    prob = pipeline.predict_proba(sample_df)[:,1][0] if hasattr(pipeline, "predict_proba") else None
    print("\n--- Result ---")
    print("Prediction:", "Fraud" if pred==1 else "Not Fraud")
    if prob is not None:
        print("Fraud Probability: {:.2f}".format(prob))
    return pred, prob


predict_from_input(best_pipeline)
```
